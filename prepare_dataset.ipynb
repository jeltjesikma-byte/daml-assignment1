```python
import os

import pandas as pd

draft = pd.read_csv("metadata_draft.csv")
draft
```

```python
draft['file'] = "data/" + draft['id'] + ".jpg"
draft['source'] = 'https://www.rijksmuseum.nl/en/collection/' + draft['id']
draft['licence'] = 'Public Domain'
draft['number'] = draft['id'].str.extract(r'(\d+)')
print(draft.head())
```

```python
draft = draft[['number', 'id', 'file', 'title', 'artist', 'date', 'source', 'licence']]
print(draft.head())
```

```python
draft.loc[7, 'date'] = 'c. 1668 - c. 1670'
print(draft.loc[7])
```

```python
draft = draft.rename(columns={'date': 'original_date'})
```

```python
draft = draft[['number', 'id', 'file', 'title', 'artist', 'original_date', 'source', 'licence']]
```

```python
draft.to_csv('metadata.csv', index=False)
```

```python
draft['earliest_date'] = (
    draft['original_date']
    .str.extract(r'(\d{4})')[0]
    .astype('Int64')
)

draft['latest_date'] = (
    draft['original_date']
    .str.extract(r'(\d{4})(?!.*\d{4})')[0]
    .astype('Int64')
)
```

```python
draft.head()
```

```python
draft = draft[['number', 'id', 'file', 'title', 'artist', 'original_date', 'earliest_date', 'latest_date', 'source', 'licence']]
```
