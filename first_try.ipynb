```python
#this code was made with help by chatGPT.

import re

def convert_edtf(date):
    if pd.isna(date):
        return date

    date = date.strip().replace('ca.', 'c.')

    # Circa-periode, bijvoorbeeld c. 1653 - 1657
    match = re.fullmatch(r'c\.\s*(\d{4})\s*-\s*c?\.\s*(\d{4})', date)
    if match:
        return f'{match.group(1)}~/{match.group(2)}~'

    # Circa-periode, bijvoorbeeld c. 1653 - 1657
    match = re.fullmatch(r'c\.\s*(\d{4})\s*-\s*(\d{4})', date)
    if match:
        return f'{match.group(1)}~/{match.group(2)}'

    # Gewone periode
    match = re.fullmatch(r'(\d{4})\s*-\s*(\d{4})', date)
    if match:
        return f'{match.group(1)}/{match.group(2)}'

    # Circa-jaartal
    match = re.fullmatch(r'c\.\s*(\d{4})', date)
    if match:
        return f'{match.group(1)}~'

    return date

df['date_edtf'] = df['original_date'].apply(convert_edtf)
```
